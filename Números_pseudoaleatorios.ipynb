# Números pseudoaleatorios

```python
# m - modulo
# a - multiplicador
# b - incremento
# X0 - semilla

def generar_pseudoaleatorios(semilla, cantidad, a=5, b=1, m=9):
    actual = semilla
    resultados = []
    for _ in range(cantidad):
        actual = (a * actual + b) % m
        resultados.append(actual)
    return resultados

print(generar_pseudoaleatorios(semilla=21, cantidad=3))
```

```python
print(generar_pseudoaleatorios(semilla=21, cantidad=30))
```

El valor máximo sería 8, porque el residuo de la división (modulo) el mayor valor sería 9.

```python
print(generar_pseudoaleatorios(semilla=7, cantidad=50, a=5, b=3, m=16))
```

El valor máximo sería 15, porque el residuo de la división (modulo) el mayor valor sería 15.
